```python
# Александрова Екатерина ЦИБ-251
# Определите значение:
# Подключаем модуль math для функции math.sqrt(abs(x))
import math


# Создаем функцию sign
def sgn(a):
  # функция возвращает знак числа: 1 - если число положительное, 0 - если число равно 0, -1 - если число отрицательное
    if a > 0:
        return 1
    elif a == 0:
        return 0
    else:
        return -1

# Создаем функцию для вычисления z
def sgn_z(x, y):
# Функция вычисляет значение выражения:
    z = (sgn(x) + y ** 2) / (sgn(y) - math.sqrt(abs(x)))

#Параметры: x - вещественное число, y - вещественное число
#Возвращает: значение z

    # Вычисляем знаменатель
    znamenatel = sgn(y) - math.sqrt(abs(x))

    # Проверяем деление на ноль
    if znamenatel == 0:
        return None
    else:
        z = (sgn(x) + y ** 2) / znamenatel
    # Возвращаем результат вычисления
    return z

#Вводим x
x = float(input())
#Вводим y
y = float(input())

#Вызываем функцию вычисления
result = sgn_z(x, y)

#Проверяем результат
if result is None:
    print("Ошибка: x и y должны быть числами!")
else:
    # Выводим результат с двумя знаками после запятой
    print("Введите x:", x)
    print("Введите y:", y)
    print(f"Ответ: {result:.2f}")
```
